```python
df.groupby('Category').agg(
    Total_Sales=('Amount', 'sum'),
    Total_Qty=('Qty', 'sum')
).assign(
    Avg_Amount_Per_Unit=lambda x: x['Total_Sales'] / x['Total_Qty']
).sort_values('Avg_Amount_Per_Unit', ascending=False)
```

```python
status_pct = df['Status'].value_counts(normalize=True) * 100
status_pct.round(2)
```

```python
df[df['Status'] == 'Cancelled']['Category'].value_counts()
```

```python
cancel_rate = (
    df.groupby('Category')['Status']
      .apply(lambda x: (x == 'Cancelled').mean() * 100)
      .sort_values(ascending=False)
)

cancel_rate.round(2)
```

```python
df.groupby('ship-state')['Amount'].sum().sort_values(ascending=False).head(10)
```

```python
df.groupby('ship-state')['Order ID'].nunique().sort_values(ascending=False).head(10)
```

```python
state_analysis = df.groupby('ship-state').agg(
    Total_Sales=('Amount', 'sum'),
    Unique_Orders=('Order ID', 'nunique')
)

state_analysis['Avg_Order_Value'] = (
    state_analysis['Total_Sales'] / state_analysis['Unique_Orders']
)

state_analysis.sort_values(
    'Avg_Order_Value',
    ascending=False
).head(10)
```

```python
state_analysis[state_analysis['Unique_Orders'] >= 500] \
    .sort_values('Avg_Order_Value', ascending=False) \
    .head(10)
```

```python
top_states = (
    df.groupby('ship-state')
      .agg(
          Total_Sales=('Amount', 'sum'),
          Unique_Orders=('Order ID', 'nunique')
      )
      .sort_values('Total_Sales', ascending=False)
      .head(10)
)
top_states
```

```python
fulfilment_sales = (
    df.groupby('Fulfilment')['Amount']
      .sum()
      .sort_values(ascending=False)
)

fulfilment_sales
```

```python
fulfilment_analysis = (
    df.groupby('Fulfilment')
      .agg(
          Total_Sales=('Amount', 'sum'),
          Unique_Orders=('Order ID', 'nunique')
      )
)

fulfilment_analysis['Avg_Order_Value'] = (
    fulfilment_analysis['Total_Sales'] /
    fulfilment_analysis['Unique_Orders']
)

fulfilment_analysis
```
